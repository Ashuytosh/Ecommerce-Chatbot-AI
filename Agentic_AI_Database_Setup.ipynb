```python
# Dictionary of all datasets
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "products": products,
    "sellers": sellers,
    "category_translation": p_category_name_translation
}

# Loop through each dataset and display dtypes
for name, df in datasets.items():
    print(f"\n {name.upper()} DATA TYPES:")
    print(df.dtypes)
```

```python
# === 2) Cleaning per table ===

## customers — keep as-is
customers_clean = customers.copy()

## geolocation — just drop exact duplicates, keep all detail
geolocation_clean = geolocation.drop_duplicates().reset_index(drop=True)

## orders — convert timestamps, keep missing (they are meaningful)
orders_clean = orders.copy()
date_cols_orders = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]
for col in date_cols_orders:
    orders_clean[col] = pd.to_datetime(orders_clean[col], errors="coerce")

## order_items — convert shipping_limit_date; ensure numeric
order_items_clean = order_items.copy()
order_items_clean["shipping_limit_date"] = pd.to_datetime(
    order_items_clean["shipping_limit_date"], errors="coerce"
)
order_items_clean["price"] = pd.to_numeric(order_items_clean["price"], errors="coerce")
order_items_clean["freight_value"] = pd.to_numeric(order_items_clean["freight_value"], errors="coerce")

## order_payments — enforce numeric types
order_payments_clean = order_payments.copy()
order_payments_clean["payment_sequential"]  = pd.to_numeric(order_payments_clean["payment_sequential"], errors="coerce").astype("Int64")
order_payments_clean["payment_installments"] = pd.to_numeric(order_payments_clean["payment_installments"], errors="coerce").astype("Int64")
order_payments_clean["payment_value"] = pd.to_numeric(order_payments_clean["payment_value"], errors="coerce")

## order_reviews — drop sparse text cols; convert dates
order_reviews_clean = order_reviews.drop(columns=["review_comment_title", "review_comment_message"])
for col in ["review_creation_date", "review_answer_timestamp"]:
    order_reviews_clean[col] = pd.to_datetime(order_reviews_clean[col], errors="coerce")

## products — drop incomplete rows; fix spelling; keep all columns
products_clean = products.copy()

# Fix typos
products_clean = products_clean.rename(
    columns={
        "product_name_lenght": "product_name_length",
        "product_description_lenght": "product_description_length",
    }
)

# Drop rows with missing critical info
required_cols = [
    "product_category_name",
    "product_name_length",
    "product_description_length",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
]
products_clean = products_clean.dropna(subset=required_cols).reset_index(drop=True)

# Ensure numeric dtypes
for c in [
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
    "product_name_length",
    "product_description_length",
]:
    products_clean[c] = pd.to_numeric(products_clean[c], errors="coerce")

## sellers — keep as-is
sellers_clean = sellers.copy()

## product_category_name_translation — keep as-is (join dynamically later)
category_translation_clean = p_category_name_translation.copy()
```

```python
# === 3) Persist to DuckDB ===
# === Database Creation Part ===
import duckdb
from pathlib import Path

# Define your dataset folder and output DB file
DATA_DIR = Path(r"D:\Agentic_AI\Dataset")
OUT_DB = DATA_DIR / "ecommerce_file.duckdb"

# Delete old database file (optional, for clean rebuild)
if OUT_DB.exists():
    OUT_DB.unlink()

# Connect to (or create) the DuckDB database
con = duckdb.connect(str(OUT_DB))

# Dictionary of all cleaned DataFrames
tables = {
    "customers": customers_clean,
    "geolocation": geolocation_clean,
    "orders": orders_clean,
    "order_items": order_items_clean,
    "order_payments": order_payments_clean,
    "order_reviews": order_reviews_clean,
    "products": products_clean,
    "sellers": sellers_clean,
    "product_category_name_translation": category_translation_clean,
}

# Register and persist each table
for name, df in tables.items():
    con.register(f"{name}_df", df)
    con.execute(f"CREATE OR REPLACE TABLE {name} AS SELECT * FROM {name}_df")

# === Optional: Create indexes for faster querying ===
con.execute("CREATE INDEX IF NOT EXISTS idx_orders_customer_id ON orders(customer_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_order_items_order_id ON order_items(order_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_order_items_product_id ON order_items(product_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_order_items_seller_id ON order_items(seller_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_products_product_id ON products(product_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_customers_customer_id ON customers(customer_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_sellers_seller_id ON sellers(seller_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_payments_order_id ON order_payments(order_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_reviews_order_id ON order_reviews(order_id)")
con.execute("CREATE INDEX IF NOT EXISTS idx_translation_cat ON product_category_name_translation(product_category_name)")

# Close the connection
con.close()

print(f"✅ All tables cleaned and saved to {OUT_DB}")
```

```python
# Testing 
import duckdb

# Connect to your saved DuckDB
con = duckdb.connect(r"D:\Agentic_AI\Dataset\ecommerce.duckdb")

# --- Example 1: Simple JOIN test ---
# Show top 5 product categories by total sales value
query = """
SELECT 
    p.product_category_name AS category,
    SUM(oi.price) AS total_sales,
    COUNT(DISTINCT oi.order_id) AS total_orders
FROM order_items oi
JOIN products p ON oi.product_id = p.product_id
GROUP BY 1
ORDER BY total_sales DESC
LIMIT 5;
"""
df = con.execute(query).fetchdf()
print(" Top 5 Categories by Total Sales:")
print(df)
```

```python
# --- Example 2: Multi-table JOIN test (customers + orders + payments) ---
# Show total payment value per state
query2 = """
SELECT 
    c.customer_state AS state,
    ROUND(SUM(op.payment_value), 2) AS total_revenue,
    COUNT(DISTINCT o.order_id) AS total_orders
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_payments op ON o.order_id = op.order_id
GROUP BY 1
ORDER BY total_revenue DESC
LIMIT 10;
"""
df2 = con.execute(query2).fetchdf()
print("\n Top 10 States by Total Revenue:")
print(df2)
```

```python
# --- Example 3: Join with review data ---
query3 = """
SELECT 
    c.customer_state,
    ROUND(AVG(r.review_score), 2) AS avg_review,
    COUNT(DISTINCT o.order_id) AS total_orders
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_reviews r ON o.order_id = r.order_id
GROUP BY 1
ORDER BY avg_review DESC
LIMIT 10;
"""
df3 = con.execute(query3).fetchdf()
print("\n Average Review Score by State:")
print(df3)

```
